```python
#import library
import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score,confusion_matrix, ConfusionMatrixDisplay
```

```python
#2. Load the dataset
data= load_breast_cancer()
x=pd.DataFrame(data.data,columns=data.feature_names)
y=pd.Series(data.target)
y
```

```python
#3. Split into training and testing sets(80% train,20% test)
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
# 4 Feature scaling (very important for KNN, because it uses Distance)
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
#5. Create and Train the KNN model
knn=KNeighborsClassifier(n_neighbors=5,metric="euclidean")
knn.fit(x_train,y_train)
```

```python
#6.Preadict on test data
y_pred=knn.predict(x_test)
```

```python
#7.Evaluate the model
accuracy=accuracy_score(y_test,y_pred)
print(f"Accuracy: {accuracy*100:.2f}%")
```

```python
knn.predict(x_test[[0]])
```

```python
cm=confusion_matrix(y_test,y_pred)
ConfusionMatrixDisplay(cm).plot()
```

```python
from sklearn.datasets import load_iris
```

```python
df=load_iris()
xx=pd.DataFrame(df.data,columns=df.feature_names)
yy=pd.Series(df.target)
xx
```

```python
print(f"Shape of Data: {xx.shape}")
print(f"Classes: {df.target_names}")
print(f"count of each class:\n {yy.value_counts()}")
```

```python
#3. Split into training and testing sets(80% train,20% test)
xx_train,xx_test,yy_train,yy_test=train_test_split(xx,yy,test_size=0.2,random_state=42,stratify=yy)
```

```python
# 4 Feature scaling (very important for KNN, because it uses Distance)
scaler=StandardScaler()
xx_train=scaler.fit_transform(xx_train)
xx_test=scaler.transform(xx_test)
```

```python
#5. Create and Train the KNN model
knnn=KNeighborsClassifier(n_neighbors=5,metric="euclidean")
knnn.fit(xx_train,yy_train)
```

```python
#6.Preadict on test data
yy_pred=knnn.predict(xx_test)
```

```python
#7.Evaluate the model
accuracy=accuracy_score(yy_test,yy_pred)
print(f"Accuracy: {accuracy*100:.2f}%")
```

```python
cmm=confusion_matrix(yy_test,yy_pred)
ConfusionMatrixDisplay(cmm).plot()
```
